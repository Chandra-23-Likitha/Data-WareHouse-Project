```python
from pyspark.sql.functions import col,trim,upper,initcap,lit
#clean crm custpmers
customer_silver=(
    spark.table("retailworks.bronze.crm_customers_raw")
    .filter(col("cust_id").isNotNull())
    .withColumn("cust_name",initcap(trim(col("cust_name"))))
    .withColumn("email",trim(col("email")))
    .withColumn("city",initcap(trim(col("city"))))
    .dropDuplicates(["cust_id"])
)
customer_silver.write.format("delta").mode("overwrite").saveAsTable("retailworks.silver.customers_clean")


```

```python
%sql
select * from retailworks.silver.customers_clean
```

```python
#clean ERP products
products_silver=(
    spark.table("retailworks.bronze.erp_products_raw")
    .filter(col("prod_id").isNotNull())
    .filter(col("unit_price")>=0)
    .withColumn("prod_name",trim(col("prod_name")))
    .dropDuplicates(["prod_id"])
)

products_silver.write.format("delta").mode("overwrite").saveAsTable("retailworks.silver.products_clean")
```

```python
%sql
select * from retailworks.silver.products_clean
```

```python
#clean orders
orders_silver=(
    spark.table("retailworks.bronze.orders_raw")
    .filter(col("quantity")>0)
    .dropDuplicates(["order_id"])
    
)
orders_silver.write.format("delta").mode("overwrite").saveAsTable("retailworks.silver.orders_clean")
```

```python
%sql
select * from retailworks.silver.orders_clean
```
