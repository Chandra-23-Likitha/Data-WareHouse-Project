```python
# Referential integrity check — every fact row should map to a valid dimension
from pyspark.sql.functions import col
orphan_customers= (
    spark.table("retailworks.gold.fact_sales").select("cust_id").distinct()
    .subtract(spark.table("retailworks.gold.dim_customer").select("cust_id").distinct())
    )
assert orphan_customers.count() == 0,"Found fact rows with no matching customer dimension!"
#Completeness Check
null_check=spark.table("retailworks.gold.fact_sales").filter(col("total_amount").isNull()).count()
assert null_check==0,"Found NULL total_amount values in fact tables!"

print("All data quality checks passed.")
```
